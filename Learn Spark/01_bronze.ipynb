#### Lendo csv

```python
# data set de exemplos do databricks
display(dbutils.fs.ls("/databricks-datasets"))
```

```python
file = "dbfs:/databricks-datasets/flights/"
```

```python
df = spark \
    .read.format("csv") \
    .option("inferSchema", True) \
    .option("header", True) \
    .csv(file)

##inferSchema = inferir o schema (valores de cada coluna) -> se for False, teremos que passar o schema manualmente
```

###### CSV -> Delta

```python
spark.sql("CREATE DATABASE IF NOT EXISTS projeto")
```

```python
display(spark.sql("show databases"))
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("projeto.bronze_flights")
```

#### Consultando dados do dataframe

```python
from pyspark.sql.functions import max
```

```python
df.select(max("delay")).take(1)
```

```python
df.filter(df.delay < 2).show()
```

```python
df.where("delay < 2").show(2)
```

```python
df.sort("delay").show(5)
```

```python
from pyspark.sql.functions import desc
df.orderBy(desc("delay")).show(5)
```
