```python
silver_df = spark.table("workspace.projeto.silver_flights")
```

##### Destino mais popular

```python
silver_df.createOrReplaceTempView("gold_flights")
```

```python
gold_top_destinos = spark.sql("""
    SELECT
        destination,
        COUNT(*) AS total_voos
    FROM
        gold_flights
    GROUP BY
        destination
    ORDER BY
        total_voos DESC
    LIMIT 3
""")

display(gold_top_destinos)
```

##### Rota de destino mais popular

```python
from pyspark.sql import functions as F

gold_rota_popular = silver_df.groupBy("origin", "destination") \
    .agg(F.count("*").alias("total_voos")) \
    .orderBy(F.col("total_voos").desc()) \
    .limit(3)

display(gold_rota_popular)
```

##### Qual a rota mais destante

```python
from pyspark.sql import functions as F

gold_rota_distante_com_count = silver_df.groupBy("origin", "destination", "distance") \
    .agg(F.count("*").alias("total_voos")) \
    .orderBy(F.col("distance").desc()) \
    .limit(3)

display(gold_rota_distante)
```
